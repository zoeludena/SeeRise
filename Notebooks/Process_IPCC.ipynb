```python
import pandas as pd
import numpy as np
```

```python
ipcc = pd.read_csv("ipcc_ar6_slr_ssp_projections.csv")

ipcc
```

```python
for i in [5, 17, 50, 83, 95]:
    med_q = ipcc[(ipcc['confidence'] == 'medium') & (ipcc['quantile'] == i)]
    med_q = med_q.drop(columns=['process', 'confidence', 'quantile']).set_index('scenario').rename(columns={'scenario': 'year'}).T
    med_q.index.names = ['year']

    med_q.to_csv(f'{i}_quantile_SLR.csv')

    df = pd.read_csv(f'{i}_quantile_SLR.csv')
    df.loc[-1] = [2010, 0, 0, 0, 0, 0]
    df = df.sort_index()
    df.index = df.index + 1
    df = df.diff().dropna().loc[1:9] / 10
    df = df.loc[np.repeat(df.index.values, 10)]
    df['year'] = np.arange(2011, 2101)
    df = df.reset_index(drop=True)

    df.to_csv(f'{i}_quantile_yearly.csv', index=False)
```
